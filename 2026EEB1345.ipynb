- The **second smallest value**.
- The **second largest value**.
- The **number of times the second smallest value occurs** in the list.
- The **number of times the second largest value occurs** in the list.
- The **index position(s)** of the second smallest value.
- The **index position(s)** of the second largest value.
- Check whether the list contains **duplicate values**.

### **Example**

**Input:**

```text
[10, 4, 8, 15, 2, 20, 4, 15]

Output

Second smallest = 4
Second largest = 15

Count of second smallest = 2
Count of second largest = 2

Positions of second smallest = [1, 6]
Positions of second largest = [3, 7]

Duplicate values = Yes

````python
liston = []
for i in range (0,8):
    liston.append(int(input("Enter your number")))

def get_second_smallest(numbers):

    smallest = float('inf')
    second_smallest = float('inf')
    
    for num in numbers:
        if num < smallest:
            second_smallest = smallest
            smallest = num
        elif num < second_smallest and num != smallest:
            second_smallest = num
            
    return second_smallest if second_smallest != float('inf') else None

def get_second_largest(numbers):

    largest = float('-inf')
    second_largest = float('-inf')
    
    for num in numbers:
        if num > largest:
            second_largest = largest
            largest = num
        elif num > second_largest and num != largest:
            second_largest = num
            
    return second_largest if second_largest != float('-inf') else None

def indx(x,y):
    st = 0
    t = 0
    f = []
    for i in (0,t):
        t = y.index(x, st)
        st = t + 1
        f.append(t)
    return f

dulp  = len(liston) != len(set(liston))

print(f"Second Smallest  = {get_second_smallest(liston)}")
print(f"Second Largest  = {get_second_largest(liston)}")
print(f"Count of the second smallest is {liston.count(get_second_smallest(liston))}")
print(f"Count of the second largest is {liston.count(get_second_smallest(liston))}")
print(f"Position of second smallest is {indx(get_second_smallest(liston),liston)}")
print(f"Position of second largest is {indx(get_second_largest(liston),liston)}")
print(f"Duplicates : {dulp}")
````
